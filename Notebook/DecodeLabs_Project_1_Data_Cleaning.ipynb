2.2 - Handling the missing values in the CouponCode column

```python
df['CouponCode'] = df['CouponCode'].fillna('Unknown')  #fill missing coupon values with 'unknown' because the dataset itself doesn't explicitly tell us that NaN means "the customer did not use a coupon."
```

```python
print(df['CouponCode'].isnull().sum()) #Verify that there are no missing values left
```

```python
df['CouponCode'].value_counts()
```

The dataset was checked for missing values using Pandas. The analysis identified 309 missing values in the `CouponCode` column. All other columns contained no missing values.

The missing `CouponCode` values were investigated by examining the affected records and the unique values in the column. Since the dataset did not provide enough information to determine whether a missing coupon represented "no coupon" or unavailable information, the missing values were replaced with `Unknown`.

After the replacement, the `CouponCode` column was verified to contain zero missing values.

5.1 - Check whether TotalPrice is calculated correctly

```python
print("Number of TotalPrice that are not calculated correctlly: ",(df['Quantity'] * df['UnitPrice'] != df['TotalPrice']).sum())
```

5.2 - Inspect the 107 mismatched records

```python
mismatched = df[
    (df['Quantity'] * df['UnitPrice']).round(2) != df['TotalPrice'].round(2)
]

mismatched.head(20)
```

```python
mismatched[['Quantity', 'UnitPrice', 'CouponCode', 'TotalPrice']].head(20)
```

```python
print((df['Quantity'] * df['UnitPrice']).round(2).eq(df['TotalPrice'].round(2)).all())
```

The `TotalPrice` column was validated against the expected calculation:

`Quantity × UnitPrice`

An initial direct comparison identified 107 apparent mismatches. However, after rounding both calculated and recorded values to two decimal places, all records matched.

This indicated that the initial differences were caused by floating-point precision rather than actual data inconsistencies.

The final validation returned `True`, confirming that all `TotalPrice` values were consistent with `Quantity × UnitPrice` to two decimal places.

No changes were made to the `TotalPrice` column.

6. Text Data Validation

6.1 - Check the text/categorical columns

```python
text_columns = [
    'Product',
    'PaymentMethod',
    'OrderStatus',
    'CouponCode',
    'ReferralSource'
]

for col in text_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

6.2 - Check for hidden spaces

```python
for col in text_columns:
    spaces = df[col].astype(str).str.strip().ne(df[col].astype(str))
    print(f"{col}: {spaces.sum()} values with leading/trailing spaces")
```

Text Format Validation: The categorical columns were checked for inconsistent capitalization, spelling variations, and leading/trailing spaces. No formatting inconsistencies were identified, so no additional text transformations were required.
